```python
# KS AI BUSINESS ASSISTANT
# Working business dashboard prototype

import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Brand colors
BROWN = "#3D2B1F"
MINT = "#A8E6CF"

title = HTML("""
<div style="background:#3D2B1F;padding:24px;border-radius:14px;color:white">
<h1>KS AI Business Assistant</h1>
<p>Smart Business Planning & Marketing Studio</p>
<p>Built by KS Designs</p>
</div>
""")

business = widgets.Text(
    description="Business:",
    placeholder="e.g. KS Designs",
    layout=widgets.Layout(width="95%")
)

industry = widgets.Dropdown(
    description="Industry:",
    options=[
        "Graphic Design",
        "Fashion",
        "Food & Restaurant",
        "Education",
        "Technology",
        "Beauty & Wellness",
        "Other"
    ],
    layout=widgets.Layout(width="95%")
)

audience = widgets.Text(
    description="Audience:",
    placeholder="e.g. Small business owners",
    layout=widgets.Layout(width="95%")
)

goal = widgets.Dropdown(
    description="Goal:",
    options=[
        "Get more customers",
        "Increase brand awareness",
        "Generate leads",
        "Improve social media",
        "Launch a new product"
    ],
    layout=widgets.Layout(width="95%")
)

task = widgets.Dropdown(
    description="Tool:",
    options=[
        "Marketing Ideas",
        "Instagram Caption",
        "LinkedIn Post",
        "Customer Reply",
        "7-Day Action Plan"
    ],
    layout=widgets.Layout(width="95%")
)

details = widgets.Textarea(
    description="Details:",
    placeholder="Describe your product, offer or customer question",
    layout=widgets.Layout(width="95%", height="100px")
)

button = widgets.Button(
    description="Generate Business Output",
    button_style="success",
    layout=widgets.Layout(width="95%", height="45px")
)

result = widgets.Output()

def generate_output(_):
    with result:
        clear_output()

        name = business.value.strip()
        market = audience.value.strip()
        extra = details.value.strip()
        selected_goal = goal.value
        selected_task = task.value

        if not name or not market:
            print("Please enter your business name and target audience.")
            return

        if selected_task == "Marketing Ideas":
            answer = f"""
MARKETING PLAN — {name}

Target audience: {market}
Business goal: {selected_goal}

1. Publish useful tips related to your industry.
2. Show examples of your work or product.
3. Explain how your service solves a customer problem.
4. Share a customer FAQ post.
5. Add a clear call to action to each post.

Next action: Publish one useful post and contact
potential customers who may need your service.
"""

        elif selected_task == "Instagram Caption":
            answer = f"""
INSTAGRAM CAPTION

Looking for better solutions for your business?

{name} helps {market} explore ideas and solutions
for their business needs.

Interested in learning more? Send us a DM!

#SmallBusiness #BusinessGrowth #KSDesigns
"""

        elif selected_task == "LinkedIn Post":
            answer = f"""
LINKEDIN POST

Every business needs practical ways to grow.

At {name}, we are exploring solutions for {market}.

Our current focus: {selected_goal}.

We believe that understanding customer problems
is the first step toward building useful solutions.

What is the biggest challenge your business faces
today?

#Business #Innovation #Entrepreneurship
"""

        elif selected_task == "Customer Reply":
            answer = f"""
CUSTOMER REPLY TEMPLATE

Hi! Thank you for your interest in {name}.

We would love to understand your requirements
and recommend a suitable solution.

Could you please share a few details about
what you are looking for?

We will review your requirements and get back
to you with the next steps.

Best regards,
{name}
"""

        else:
            answer = f"""
7-DAY BUSINESS ACTION PLAN — {name}

Day 1: Identify your ideal customers.
Day 2: Prepare a sample of your service.
Day 3: Publish a useful educational post.
Day 4: Contact relevant potential customers.
Day 5: Demonstrate your service to interested people.
Day 6: Collect feedback and improve your offer.
Day 7: Follow up with interested prospects.

Goal: {selected_goal}
Target audience: {market}

Note: Results are not guaranteed.
"""

        if extra:
            answer += "\n\nYOUR ADDITIONAL DETAILS:\n" + extra

        print(answer)

        copy_button = widgets.Button(
            description="Copy Output",
            button_style="info"
        )

        def copy_output(_):
            from IPython.display import Javascript, display
            safe_text = answer.replace("\\", "\\\\")
            safe_text = safe_text.replace("`", "\\`")
            display(Javascript(
                "navigator.clipboard.writeText(" +
                repr(answer) + ").then(() => alert('Copied!'))"
            ))

        copy_button.on_click(copy_output)
        display(copy_button)

button.on_click(generate_output)

display(title)
display(widgets.HTML("<h3>Business Setup</h3>"))
display(business, industry, audience, goal)
display(widgets.HTML("<h3>Choose Your AI Tool</h3>"))
display(task, details, button)
display(result)
```
